```python
x = data.loc[:,'age':'time']
y = data.loc[:,["DEATH_EVENT"]]

x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=2)
```

Selects features and the target variable, then splits the dataset into 80% training data and 20% testing data.

```python
gradientboost_clf = GradientBoostingClassifier(max_depth=2)
gradientboost_clf.fit(x_train,y_train)
gradientboost_pred = gradientboost_clf.predict(x_test)
```

Creates a Gradient Boosting classifier, trains it on the training data, and predicts outcomes for the test data.

```python
gradientboost_clf.score(x_test,y_test)
```

Calculates the accuracy of the Gradient Boosting model on the test data.

```python
from sklearn.metrics import confusion_matrix
cf_matrix = confusion_matrix(y_test, gradientboost_pred)
```

Imports the confusion matrix function and creates a confusion matrix to evaluate the model's prediction performance.

```python
sns.heatmap(cf_matrix, cmap='Blues',annot = True)
```
